```python
#3.1
import numpy as np
from scipy import sparse

filename = "test.dat"
rng = np.random.default_rng(42)
m = np.memmap(filename, dtype="float64", mode="w+", shape=(1000, 1000))
for i in range(1000):
    m[i] = rng.random(1000)
m.flush()
print("Файл создан ", filename)
print("Форма ", m.shape, "Размер = ", m.nbytes / 1024**2, "МиБ")
```

```python
#3.2
m = np.memmap(filename, dtype="float64", mode="r", shape=(1000, 1000))
row = m[10, :]
column = m[:, 10]
print("Строка 10, первые 5 элементов ", row[:5])
print("Столбец 10, первые 5 элементов ", column[:5])
```

```python
#3.3
def b():
    total = 0.0
    for i in range(1000):
        total += np.sum(m[i, :])
    return total

def c():
    total = 0.0
    for j in range(1000):
        total += np.sum(m[:, j])
    return total

print("Сумма по строкам ", b())
print("Сумма по столбцам ", c())
```

```python
#4.1
small = np.zeros((5, 5))
small[0, 1] = 2
small[1, 3] = 4
small[2, 0] = 6
small[3, 4] = 8
small[4, 2] = 10
csr = sparse.csr_matrix(small)
print("Матрица \n", csr.toarray())
print("Ненулевых элементов ", csr.nnz)
print("Плотность ", csr.nnz / 25)
```

```python
#4.2
coo = csr.tocoo()
csc = csr.tocsc()
print("Исходный формат ", csr.getformat())
print("COO ", coo.getformat())
print("CSC ", csc.getformat())
print("Содержимое COO \n", coo.toarray())
print("Содержимое CSC \n", csc.toarray())
```

```python
#4.3
v = np.array([1, 2, 3, 4, 5])
print("Вектор ", v)
print("Результат =", csr @ v)
```

```python
#4.4
large_csr = sparse.random(1000, 1000, density=0.01, format="csr", random_state=42)
large_dense = large_csr.toarray()
vector = np.ones(1000)
print("Ненулевых элементов ", large_csr.nnz)
print("Результаты совпадают ", np.allclose(large_csr @ vector, large_dense @ vector))
```

```python
#5.1
def sum_squares(values):
    t = 0.0
    for x in values:
        t += x * x
    return t

print("Проверка Python:", sum_squares([1.0, 2.0, 3.0]))
```

```python
#5.2
%%writefile cython_squares.pyx
def sum_squares_c(list values):
    cdef Py_ssize_t i, n = len(values)
    cdef double t = 0.0
    cdef double x
    for i in range(n):
        x = values[i]
        t += x * x
    return t
```

```python
import pyximport
pyximport.install(language_level=3)
from cython_squares import sum_squares_c
print("Проверка Cython:", sum_squares_c([1.0, 2.0, 3.0]))
```

```python
#5.3
values = [float(i % 100) for i in range(10_000_000)]
py_result = sum_squares(values)
cy_result = sum_squares_c(values)
print("Результат Python ", py_result)
print("Результат Cython ", cy_result)
print("Результаты совпадают ", np.isclose(py_result, cy_result))

%timeit -n 1 -r 3 sum_squares(values)
%timeit -n 1 -r 3 sum_squares_c(values)
```
